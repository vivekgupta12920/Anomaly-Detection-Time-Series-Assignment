Q1: What is Anomaly Detection? Explain its types (point, contextual, and
collective anomalies) with examples.


-> Anomaly Detection—also known as outlier analysis—is a core data mining and machine learning technique used to identify data points, events, or observations that deviate significantly from a dataset's normal, established baseline. These unusual patterns often signal critical incidents such as fraudulent transactions, system failures, cybersecurity breaches, or medical issues.


1. Point Anomalies (Global Outliers)
A point anomaly occurs when a single instance of data stands out as significantly different from the entire dataset. These are the simplest and most common type of anomalies to detect because the individual point falls far outside the normal distribution.
Example: A credit card purchase of $10,000 for a customer whose typical monthly bill is around $2,000.
Example: A sudden, one-time spike in CPU usage to 90% when the average is 30%.

2. Contextual Anomalies (Conditional Outliers)
A contextual anomaly is a data point that is considered abnormal only in a specific context (such as time, location, or season), even if the value itself might appear normal in a different setting.
Example: A temperature reading of 85°F (30°C) is normal during a summer heatwave but would be flagged as a severe anomaly in the middle of a New York winter.
Example: An unusually high volume of website traffic at 3:00 AM might be normal for a global e-commerce site but suspicious for a small local business.

3. Collective Anomalies
A collective anomaly occurs when a group of related data points behaves unusually when viewed as a whole, even though individual points within the group may not be anomalous on their own. These patterns emerge only when observing sequences or clusters of data.

Q3: What are the key components of a Time Series? Explain each with one
example.


-> 1. Trend
The Trend represents the long-term movement or general direction of the data over a significant period. It shows whether the values are generally increasing, decreasing, or remaining stagnant, ignoring short-term fluctuations.
Example: The steady increase in Global Average Temperature over the last 50 years. While individual years might be cooler than the previous ones, the multi-decade trend is upward.

2. Seasonality
Seasonality refers to periodic fluctuations that repeat at regular, fixed intervals (e.g., daily, weekly, monthly, or quarterly). These patterns are usually tied to the calendar or clock.
Example: Retail sales for winter coats, which consistently spike every November and December and drop significantly every July, regardless of the year.

3. Cyclical Patterns
Unlike seasonality, Cyclical components are fluctuations that do not have a fixed period. They represent rises and falls that occur over longer durations (often years) and are usually influenced by external macro-economic or physical factors.
Example: The Economic Business Cycle, where a country moves through periods of expansion, peak, recession, and recovery over a span of 5 to 10 years.

4. Irregularity (Noise)
Also known as Residuals or Noise, this component represents the random, unpredictable variations in the data. These are fluctuations that cannot be explained by trend, seasonality, or cycles and are often caused by one-off events.
Example: A sudden, massive drop in Stock Market prices caused by an unexpected natural disaster or a geopolitical crisis.




Q4: Define Stationary in time series. How can you test and transform a
non-stationary series into a stationary one?


-> There are two primary forms of stationarity:

Weak (Second-Order) Stationarity: The most common practical standard. It requires a constant mean and variance over time, and a covariance between any two points that depends only on the time lag between them.

Strict Stationarity: A more rigorous (and rare) condition where the entire probability distribution of the data remains identical regardless of when the observations were recorded.


Visual Inspection:
Time Plot: Check if the series fluctuates around a constant horizontal line (mean) with stable amplitude (variance).
ACF/PACF Plots: In stationary series, the Autocorrelation Function (ACF) typically decays to zero rapidly. Slow decay suggests non-stationarity.

Statistical (Unit Root) Tests:
Augmented Dickey-Fuller (ADF) Test: Tests the null hypothesis (\(H_{0}\)) that the series is non-stationary (contains a unit root). A \(p\text{-value}\le 0.05\) indicates stationarity.
Kwiatkowski-Phillips-Schmidt-Shin (KPSS) Test: Flips the hypothesis; the null (\(H_{0}\)) assumes the series is stationary. A \(p\text{-value}>0.05\) suggests stationarity.Best Practice: Use both tests. If they disagree, the series may be "trend-stationary" (stationary only after a trend is removed) or "difference-stationary" (requires differencing).



Q5: Differentiate between AR, MA, ARIMA, SARIMA, and SARIMAX models in
terms of structure and application.


-> 1. AR (Autoregressive) Structure:
 Uses a linear combination of its own previous values (\(y_{t-1},y_{t-2},\dots \)) to predict the current value.
 It assumes the past is a good predictor of the future.Application: Ideal for processes with inertia, such as daily temperature where today's weather is highly dependent on yesterday's.

 2. MA (Moving Average)
Structure: Predicts the current value based on past "shocks" or forecast errors rather than the raw values themselves. It essentially uses a moving window of past residuals to smooth noise.
Application: Useful for identifying and "correcting" for unexpected past errors or sudden short-term spikes in data.


3. ARIMA (Autoregressive Integrated Moving Average)
Structure: Combines AR and MA components with an Integrated (\(I\)) component. The \(I\) component uses differencing to make non-stationary data stationary by removing trends.
Application: A standard "baseline" for datasets that have a clear upward or downward trend but no repeating seasonal cycles.


4. SARIMA (Seasonal ARIMA) Structure: Adds four seasonal parameters \((P,D,Q,s)\) to the standard ARIMA model.
This allows it to handle patterns that repeat every \(s\) time steps (e.g., \(s=12\) for monthly data).
Application: Essential for data with recurring cycles, such as quarterly financial results or energy consumption that peaks every summer.


5. SARIMAX (SARIMA with Exogenous Regressors)
 Structure: The most advanced classical model. It includes everything in SARIMA but adds exogenous variables (\(X\))—external data points that are not part of the time series itself but influence it.
Application: High-precision forecasting where outside factors matter. For example, predicting ice cream sales (target) using both past sales and external temperature as an exogenous regressor.



Q6: Load a time series dataset (e.g., AirPassengers), plot the original series,
and decompose it into trend, seasonality, and residual components


```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# 1. Load the AirPassengers dataset
# Most 2026 environments use this standard URL for the dataset
url = "raw.githubusercontent.com"
df = pd.read_csv(url)

# 2. Prepare the data
# Convert 'Month' to datetime and set it as the index
df['Month'] = pd.to_datetime(df['Month'])
df.set_index('Month', inplace=True)

# 3. Decompose the series
# We use 'multiplicative' because variance increases over time
decomposition = seasonal_decompose(df['Passengers'], model='multiplicative', period=12)

# 4. Extract components
trend = decomposition.trend
seasonal = decomposition.seasonal
residual = decomposition.resid

# 5. Visualize the results
plt.figure(figsize=(10, 8))

plt.subplot(411)
plt.plot(df['Passengers'], label='Original (Observed)', color='blue')
plt.legend(loc='upper left')
plt.title('Time Series Decomposition: AirPassengers')

plt.subplot(412)
plt.plot(trend, label='Trend', color='orange')
plt.legend(loc='upper left')

plt.subplot(413)
plt.plot(seasonal, label='Seasonality', color='green')
plt.legend(loc='upper left')

plt.subplot(414)
plt.scatter(df.index, residual, label='Residuals (Noise)', color='red', s=10)
plt.axhline(y=1, color='black', linestyle='--') # 1 is the baseline for multiplicative
plt.legend(loc='upper left')

plt.tight_layout()
plt.show()
```

Q7:  Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to
detect anomalies. Visualize the anomalies on a 2D scatter plot.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# 1. Load a subset of the NYC Taxi Fare dataset
# For 2026 workflows, ensure distance and fare_amount are cleaned
url = "raw.githubusercontent.com"
df = pd.read_csv(url)

# 2. Select numerical features for anomaly detection
X = df[['distance', 'fare_amount']]

# 3. Apply Isolation Forest
# 'contamination' sets the expected % of outliers (e.g., 2%)
model = IsolationForest(contamination=0.02, random_state=42)
df['anomaly_label'] = model.fit_predict(X)

# 4. Extract inliers (1) and outliers (-1)
inliers = df[df['anomaly_label'] == 1]
outliers = df[df['anomaly_label'] == -1]

# 5. Visualize on a 2D scatter plot
plt.figure(figsize=(10, 6))

# Plot normal points in green
plt.scatter(inliers['distance'], inliers['fare_amount'],
            c='green', label='Normal', alpha=0.5, s=20)

# Plot anomalies in red
plt.scatter(outliers['distance'], outliers['fare_amount'],
            c='red', label='Anomaly', alpha=0.8, s=40, marker='x')

plt.title("NYC Taxi Fare Anomaly Detection (Isolation Forest 2026)")
plt.xlabel("Trip Distance (miles)")
plt.ylabel("Fare Amount ($)")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
